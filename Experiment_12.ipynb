# Experiment 12: Basic Data Indexing for RAG

Regenerated as a standalone notebook using Google Gemini API.

```python

import os, json, re, getpass
try:
    from google import genai
except ImportError as exc:
    raise ImportError("Install the Gemini SDK first: %pip install -q google-genai") from exc

# Use a new Gemini API key. Input is hidden and not saved in this notebook.
API_KEY = os.getenv("GEMINI_API_KEY") or getpass.getpass("Enter your Gemini API key (input hidden): ")
MODEL = "gemini-2.5-flash"
client = genai.Client(api_key=API_KEY)

def ask_llm(prompt, system="You are a helpful assistant.", temperature=0.2):
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config={"system_instruction": system, "temperature": temperature}
    )
    return response.text or ""

print("Gemini client configured. Model:", MODEL)
```

## Experiment Program

```python
text_data = ("Climate change causes rising temperatures and sea levels. Melting glaciers increase coastal flooding. "
             "Extreme weather can threaten ecosystems and communities. Solar and wind energy can reduce emissions.")
def make_chunks(text, size=100, overlap=20):
    chunks, start = [], 0
    while start < len(text):
        chunks.append(text[start:start+size])
        if start + size >= len(text): break
        start += size-overlap
    return chunks
chunks = make_chunks(text_data)
query = "What are the effects of climate change?"
qwords = set(re.findall(r"\\w+", query.lower()))
ranked = sorted([(len(qwords & set(re.findall(r"\\w+", c.lower()))) / max(1,len(qwords)), c) for c in chunks], reverse=True)
print("Chunks:", len(chunks))
for score, chunk in ranked[:2]:
    print(f"Score={score:.2f} | {chunk}")
print("Optional vector store: install langchain-community, langchain-google-genai, and faiss-cpu.")
```
